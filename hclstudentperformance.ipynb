```python
# Data Analytics Project - Student Performance Analysis
# Dataset: student_performance_attendance.csv

import pandas as pd

# 1. LOAD DATASET
df = pd.read_csv("student_performance_attendance.csv")
print("--- Original Dataset ---")
print(f"Total Rows: {len(df)}")
print(df.head())

# 2. DATA CLEANING
# Duplicate cleaning
df = df.drop_duplicates(subset=["student_id"])

# Gender cleaning
df["gender"] = df["gender"].str.strip().str.lower()
df["gender"] = df["gender"].replace({"m": "Male", "male": "Male", "f": "Female", "female": "Female"})

# Clean participation column
df["participation"] = df["participation"].str.strip().str.title()

# replace invalid values with missing values,age should be between 10 and 30,marks should be between 0 and 100
df.loc[(df["age"] < 10) | (df["age"] > 30), "age"] = None
df.loc[(df["assignment_score"] < 0) | (df["assignment_score"] > 100), "assignment_score"] = None
df.loc[(df["midterm_score"] < 0) | (df["midterm_score"] > 100), "midterm_score"] = None
df.loc[(df["final_score"] < 0) | (df["final_score"] > 100), "final_score"] = None

# fill missing values with average
df = df.fillna(df.mean(numeric_only=True))

print("\n--- After Cleaning ---")
print(f"Total Rows: {len(df)}")

# 3. ANALYSIS - WHILE loop to calculate total
marks_list = df["final_score"].tolist()
attendance_list = df["attendance_percentage"].tolist()

total = 0
i = 0
while i < len(marks_list):
    total = total + marks_list[i]
    i = i + 1

average = total / len(marks_list)
highest = max(marks_list)
lowest = min(marks_list)

print("\n--- Overall Analysis ---")
print(f"Average Final Score : {average:.2f}")
print(f"Highest Final Score : {highest}")
print(f"Lowest Final Score  : {lowest}")
print(f"Average Attendance  : {sum(attendance_list)/len(attendance_list):.2f}%")

# 4. REPORT CARD - ( GRADE + ATTENDANCE )- FOR loop
print("\n--- Report Card (First 10 Students) ---")
for j in range(10):
    mark = marks_list[j]
    attend = attendance_list[j]

    if mark >= 90:
        grade = "A+"
    elif mark >= 75:
        grade = "A"
    elif mark >= 60:
        grade = "B"
    elif mark >= 40:
        grade = "C"
    else:
        grade = "Fail"

    if attend < 75:
        status = "Low Attendance!"
    else:
        status = "Good"

    print(f"{df['student_id'][j]} - Marks: {mark:.1f}, Grade: {grade}, Attendance: {attend}%, {status}")

# 5. FINAL RESULT
if average >= 60:
    print("\nOverall Performance: GOOD")
else:
    print("\nOverall Performance: NEEDS IMPROVEMENT")
```
